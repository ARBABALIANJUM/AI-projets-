```
!pip install faiss-cpu
!pip install pypdf
!pip install langchain-community
import os
from sentence_transformers import SentenceTransformer
from langchain.vectorstores import FAISS
from langchain.document_loaders import PyPDFLoader
from langchain.schema import Document
from langchain_community.embeddings import HuggingFaceEmbeddings

# Step 1: Load the local embeddings model
model_name = 'all-MiniLM-L6-v2'
model = HuggingFaceEmbeddings(model_name=model_name)

# Step 2: Load your PDF file
pdf_path = "/content/RAG Dictionary (A-Z Guide).pdf"
loader = PyPDFLoader(pdf_path)
documents = loader.load()
```

```
from langchain.text_splitter import RecursiveCharacterTextSplitter
# Step 3: Split document into smaller chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,  # Adjust chunk size as needed
    chunk_overlap=50  # Add slight overlap for context
)
chunks = text_splitter.split_documents(documents)
```

```
# Step 4: Generate embeddings and store in FAISS
db = FAISS.from_documents(chunks, model)
```

```
# Step 5: Search the vector store
query = "Contextual Embedding"

results = db.similarity_search(query)
print(results[0].page_content)
```
